# Exploring a Portfolio Risk Surface

Enbridge and TD over the last 252 trading days. Volatilities **17.4%** and **17.1%**, correlation **0.019**, safest mix **49.0% ENB** with volatility **12.3%**.

## 1. Load the summary

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent
DATA = ROOT / 'data' / 'risk_surface_outputs'
summary = json.loads((DATA / 'risk_surface_summary.json').read_text())
summary
```

## 2. The two surfaces

Value is a plane in the two prices. Volatility is a curved surface in weight and correlation. The grid files hold both.

```python
value = pd.read_csv(DATA / 'value_surface_grid.csv')
risk = pd.read_csv(DATA / 'risk_surface_grid.csv')
value.shape, risk.shape
```

## 3. Directional moves

A directional derivative prices a combined move in one step. For value it is exact, because value is linear. Note the near-offset when ENB falls 10% and TD rises 10%.

```python
pd.read_csv(DATA / 'direction_scenarios.csv')
```

## 4. Correlation scenarios

The safest weight barely moves with correlation. The level of risk moves a lot.

```python
pd.read_csv(DATA / 'correlation_scenarios.csv')
```

## 5. Taylor approximations

First order is the tangent plane. Second order adds curvature from the Hessian. Both are local tools: errors grow for large correlation changes.

```python
pd.read_csv(DATA / 'taylor_risk_approximations.csv')
```

## Conclusion

At the safest weight the volatility gradient in the weight direction is zero, as it must be at a minimum. The Hessian has one positive and one negative eigenvalue, so the joint surface is a saddle: bowl-shaped along weight and cap-shaped along correlation.
