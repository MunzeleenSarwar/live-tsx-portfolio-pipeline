# Is This Trading Strategy Actually Profitable?

Monthly 12-1 momentum on ten TSX stocks: **111 months**, annualised return **27.7%**, t-statistic **3.27**, decision **DO NOT TRADE**.

Profitable in sample is not the same as worth trading. This notebook walks through the full test.

## 1. Load the saved outputs

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent
DATA = ROOT / 'data' / 'strategy_evaluation_outputs'
summary = json.loads((DATA / 'strategy_summary.json').read_text())
monthly = pd.read_csv(DATA / 'monthly_strategy_returns.csv', parse_dates=['holding_date'])
lots = pd.read_csv(DATA / 'trade_lots.csv')
summary['decision']
```

## 2. Expected value comes before returns

A trade can win less than half the time and still be profitable if wins are larger than losses. The 333 trade lots give the payoff table.

```python
payoff = pd.read_csv(DATA / 'payoff_summary.csv')
payoff
```

## 3. Portfolio results and the equal-weight problem

The strategy beat the index. The harder test is the same ten stocks held in equal weight, which had a higher Sharpe ratio and a smaller drawdown.

```python
pd.read_csv(DATA / 'performance_metrics.csv')
```

## 4. Monte Carlo and the placebo

Bootstrap paths resample the realised months. The placebo selects random stocks each month in the same universe. Random selection did well too, which points to universe selection, not only to momentum skill.

```python
boot = pd.read_csv(DATA / 'bootstrap_summary.csv')
placebo = pd.read_csv(DATA / 'placebo_summary.csv')
boot, placebo
```

## 5. Stress tests and the decision

```python
stress = pd.read_csv(DATA / 'stress_tests.csv')
criteria = pd.read_csv(DATA / 'decision_criteria.csv')
stress[['scenario', 'annualised_return', 'sharpe', 'max_drawdown']], criteria
```

## Conclusion

The strategy passes the profitability tests and fails the risk and data-integrity tests. DO NOT TRADE with live capital. Keep it as a research strategy and a paper-trading candidate.
