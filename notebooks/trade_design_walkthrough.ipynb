# Designing an Optimal Trade

A long trade in CNQ, the rank 1 momentum stock on 30 Oct 2026, designed for a hypothetical $1,000,000 account.

Final plan: **2,551 shares** at **$78**, stop **$75**, target **$84**, risk at stop **$9,617**.

## 1. Load the plan

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent
DATA = ROOT / 'data' / 'trade_design_outputs'
plan = json.loads((DATA / 'trade_plan.json').read_text())
plan
```

## 2. Profit as a function of size

Gross expected profit rises in a straight line with size. The quadratic impact cost bends it down. On a stock this liquid the impact-only optimum is unrealistically large, which is itself the lesson.

```python
pd.read_csv(DATA / 'profit_by_size.csv')
```

## 3. Leverage and the risk penalty

Adding a variance penalty gives an interior optimum near 0.34 times capital. Raw Kelly is shown only to be rejected.

```python
lev = pd.read_csv(DATA / 'leverage_analysis.csv')
kelly = pd.read_csv(DATA / 'kelly_analysis.csv')
lev.iloc[::5], kelly
```

## 4. Stop placement

Under a driftless random walk, the probability of reaching the target before the stop depends only on the two distances. With a 1.5 reward-risk ratio the no-edge win probability is 40%, which is also the break-even probability.

```python
pd.read_csv(DATA / 'stoploss_analysis.csv')
```

## 5. Execution speed

At 2,551 shares the order is about 0.03% of one day's volume. Impact is below 1 bp even in a single day, while the timing risk of waiting grows with the square root of time. Execute within one day.

```python
pd.read_csv(DATA / 'execution_analysis.csv')
```
